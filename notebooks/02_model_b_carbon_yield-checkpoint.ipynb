```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os 
import sklearn

BASE_DIR = os.path.abspath(os.path.join(os.getcwd(), ".."))
DATA_DIR = os.path.join(BASE_DIR, "data")
DATASET_PATH = os.path.join(DATA_DIR, "model_b_agroforestry_carbon.csv")

print(f"Project Root: {BASE_DIR}")
print(f"Data Directory: {DATA_DIR}")
```

```python
df = pd.read_csv(DATASET_PATH)

print(f"Model B Dataset Loaded Successfully!")
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns\n")
df.head()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 5))
sns.histplot( data=df, x='carbon_sequestration_t_co2e_ha_yr', kde=True, ax=axes[0], color='forestgreen', bins=30)
axes[0].set_title("1. Distribution of Carbon Sequestration\n(t CO2e / ha / yr)", fontsize=11, fontweight='bold')
axes[0].set_xlabel("t CO2e / ha / yr")
axes[0].set_ylabel("Count")
top_10_species = (
    df.groupby('label')['carbon_sequestration_t_co2e_ha_yr']
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
sns.barplot(
    data=top_10_species, 
    x='carbon_sequestration_t_co2e_ha_yr', 
    y='label', 
    ax=axes[1], 
    palette="YlGn_r"
)
axes[1].set_title("2. Top 10 Species by Mean Carbon Rate", fontsize=11, fontweight='bold')
axes[1].set_xlabel("Mean t CO2e / ha / yr")
axes[1].set_ylabel("Species Label")

# 3. Correlation Heatmap: Features vs. Target Metrics
num_cols = ['n', 'p', 'k', 'temperature', 'humidity', 'ph', 'rainfall', 'biomass_yield_t_ha_yr', 'carbon_sequestration_t_co2e_ha_yr']
corr_matrix = df[num_cols].corr()

sns.heatmap(
    corr_matrix[['carbon_sequestration_t_co2e_ha_yr']].sort_values(by='carbon_sequestration_t_co2e_ha_yr', ascending=False), 
    annot=True, 
    cmap="Blues", 
    fmt=".2f", 
    ax=axes[2],
    cbar=False
)
axes[2].set_title("3. Feature Correlations\nwith Carbon Sequestration", fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()

# Print Target Summary Statistics
print("--- Model B Target Summary Statistics ---")
display(df[['biomass_yield_t_ha_yr', 'carbon_sequestration_t_co2e_ha_yr', 'time_to_maturity_years']].describe().T)
```
